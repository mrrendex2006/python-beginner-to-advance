```python

x=[1,2,3,4,5]
dsa_marks=[90,90,80,69,80]
ml_marks=[93,79,59,89,80]

plt.plot(x,dsa_marks,label="dsa marks",color= 'm',linewidth=2,marker='*',linestyle='--',markersize=10)
plt.plot(x,ml_marks,'r*--',label="ml marks ")# by using short hand notation

plt.title("marks corresonding to no of hours study",
        fontdict={'fontname':'comic as same','fontweight':10,'color':'blue'})
plt.xlabel("no of hours studied",fontdict={'fontname':'comic as same','fontweight':10,'fontsize':10,'color':'red'})
plt.ylabel("marks obtain",fontdict={'fontname':'comic as same','fontweight':10,'fontsize':20,'color':'green'})

plt.xticks([1,2,3,4,5])
plt.yticks([10,20,30,40,50])
plt.legend(loc='lower right')
plt.savefig("marks.png",dpi=10)# to save your graph ,figure ets use savefig and mension resolution br dpi
plt.show()  
```

```python
import matplotlib.pyplot as plt
import numpy as np
t=np.arange(0.,5.,0.2)
plt.figure()

plt.subplot(3,1,1)
plt.plot(t,t,'r--')
plt.title('linear')

plt.subplot(3,1,2)
plt.plot(t,t**2,'g^')
plt.title('quadratic')

plt.subplot(3,1,3)
plt.plot(t,t**3,'bs')
plt.title('qubic')

plt.suptitle('different function of t')
plt.tight_layout()# erridicats spacing issues 

plt.show()
```

```python
x=np.linspace(0,10,100)
fig , axes=plt.subplots(3,3)

axes[0,0].plot(x,np.sin(x))
axes[0,0].set_title('sin fun')
axes[0,0].set_xlabel("x")
axes[0,0].set_ylabel("y")

axes[0,1].plot(x,np.cos(x))
axes[0,1].set_title('cos fun')
axes[0,1].set_xlabel("x")
axes[0,1].set_ylabel("y")

axes[0,2].plot(x,np.tan(x))
axes[0,2].set_title('tan fun')
axes[0,2].set_xlabel("x")
axes[0,2].set_ylabel("y")

axes[1,0].plot(x,np.exp(x))
axes[1,0].set_title('exponantial fun')
axes[1,0].set_xlabel("x")
axes[1,0].set_ylabel("y")

axes[1,1].plot(x,np.sqrt(x))
axes[1,1].set_title('sqrt fun')
axes[1,1].set_xlabel("x")
axes[1,1].set_ylabel("y")

axes[1,2].plot(x,np.log(x))
axes[1,2].set_title('logt fun')
axes[1,2].set_xlabel("x")
axes[1,2].set_ylabel("y")


axes[2,0].plot(x,x)
axes[2,0].set_title('linear fun')
axes[2,0].set_xlabel("x")
axes[2,0].set_ylabel("y")

axes[2,1].plot(x,np.sqrt(x))
axes[2,1].set_title('sqrt fun')
axes[2,1].set_xlabel("x")
axes[2,1].set_ylabel("y")
plt.tight_layout()
```

SCATTER PLOT :

```python
x=np.random.rand(1000)*10
y=np.random.rand(1000)*10

plt.scatter(x,y,marker='D',edgecolors='black',linewidths=1,alpha=0.5,label='random')
plt.xlabel('people employed')
plt.ylabel('people getting retired',color='red')
plt.grid(True)
plt.legend(loc="lower right")
plt.show()
```

BAR PLOT

```python
cources=['GenAI','ML','lld','MLD','python','MLOPs']
students=[250,300,400,200,500,200]

plt.bar(cources,students,width=0.5,align='center',edgecolor='black',linewidth=3)
plt.title("enrollment of student in the cources")
plt.xlabel('cources')
plt.ylabel('no of students')
plt.grid(True)
plt.show()
```

```python
cources=['GenAI','ML','lld','MLD','python','MLOPs']
students=[250,300,400,200,500,200]
hatches=['/','*','\\','|','=','-']
color=['red','pink','magenta','yellow','green','blue']


bar=plt.bar(cources,students,width=0.5,align='center',linewidth=3)

for bar,hatches,color in zip(bar,hatches,color):
    bar.set_hatch(hatches)
    bar.set_color(color)
    bar.set_edgecolor('black')

plt.title("enrollment of student in the cources")
plt.xlabel('cources')
plt.ylabel('no of students')
plt.show()
```

HISTOGRAM PLOT:

```python
age=[20,30,28,18,34,35,40,23,44,34,32,25,27,28]
sorted_age=np.sort(age)
print(sorted_age)
plt.hist(age,bins=5,orientation='horizontal',linewidth=2,edgecolor='black')
plt.show()
```

PIE PLOT

```python
cources=['GenAI','ML','lld','MLD','python','MLOPs']
students=[250,300,400,200,500,200]
explode=[0.2,0,0,0,0,0]

plt.pie(students,labels=cources,shadow=True,explode=explode,autopct="%.2fw")
plt.legend()
plt.show()
```
